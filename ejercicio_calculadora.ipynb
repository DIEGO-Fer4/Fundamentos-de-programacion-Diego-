# Ejercicio calculadora


```python
nombre = input("¿Cómo te llamas? ")
horas_redes = float(input("Horas al día en Instagram: "))
horas_mensajeria = float(input("Horas al día en WhatsApp: "))
horas_streaming = float(input("Horas al día en streaming (series, videos): "))
horas_juegos = float(input("Horas al día en Fortnite: "))
horas_musica = float(input("Horas al día escuchando Heavy Metal: "))

tiempo_total = horas_redes + horas_mensajeria + horas_streaming + horas_juegos + horas_musica
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE MI RUTINA DIGITAL -----")
print(f"Usuario: {nombre}")
print(f"Tiempo total en pantalla: {tiempo_total} horas")
print(f"Porcentaje de tu día: {porcentaje:.2f}%")
```
