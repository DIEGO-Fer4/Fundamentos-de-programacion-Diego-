## Paso 1 — Ejecuta la calculadora (versión de 3 plataformas)

Ejecuta la celda de abajo con `Shift + Enter`. El programa te hará preguntas en un cuadro de texto: escribe tu respuesta y presiona `Enter` después de cada una. Usa **tus horas reales**. Para escribir media hora usa punto, no coma: `1.5`.

```python
nombre = input("¿Cómo te llamas? ")
horas_redes = float(input("Horas al día en Instagram: "))
horas_mensajeria = float(input("Horas al día en WhatsApp: "))
horas_streaming = float(input("Horas al día en streaming (series, videos): "))

tiempo_total = horas_redes + horas_mensajeria + horas_streaming
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE MI RUTINA DIGITAL -----")
print("Usuario", nombre)
print("Tiempo total en pantalla", tiempo_total, "horas")
print("Porcentaje de tu día:", f"{porcentaje:.2f}%")
```
