Parte B

Mi código resuleve el registro de la entrada de personas o visitantes a un museo, por lo que primero solicita el número total de integrantes a ingresar y después solicitará su edad y a qué se dedica, (profesor , estudiante o ninguno), Los bebés y personas menores a 18 años no pagan, de 18 años en adelante dependerá si es estudiante, profesor o ninguno y se le aplicará un 10% o hasta un 12% para profesores de adulto mayor de mayores de 60 años.

```python
PRECIO_CHAMACO = 0.0
PRECIO_JOVEN = 30.0
PRECIO_ADULTOS = 45.0

DESCUENTO_ADULTO_MAYOR = 0.12  
DESCUENTO_ESTUDIANTE = 0.10    
DESCUENTO_PROFESOR = 0.10     
DESCUENTO_NINGUNO = 0.0      

EDAD_MIN = 0
EDAD_MAX = 110
EDAD_JOVEN = 3
EDAD_ADULTO = 18
EDAD_MAYOR = 60

total_visitantes_cobrados = 0
total_descuentos_acumulados = 0.0
total_recaudado = 0.0

print("=== TAQUILLA DEL MUSEO ===")

while True:
    entrada_visitantes = input("¿Cuántos visitantes quieren boleto para el museo? (escribe 'fin' para salir): ")
    
    if entrada_visitantes.lower() == 'fin':
        print("La taquilla se cierra")
        break

    total_a_atender = int(entrada_visitantes)
    
    visitante_actual = 0

    while visitante_actual < total_a_atender:
        print()
        print("--- Visitante", visitante_actual + 1, "de", total_a_atender, "---")
        entrada = input("¿Cuál es la edad del visitante? (escribe 'fin' para salir): ")
        
        if entrada.lower() == 'fin':
            print("La taquilla se cierra")
            break
            
        edad = int(entrada)

        if edad < EDAD_MIN or edad > EDAD_MAX:
            print("Una edad negativa o mayor de 110 no es válida")
            continue

        if edad < EDAD_JOVEN:
            precio_base = PRECIO_CHAMACO
            tipo_visita = "Menor de 3 años"
            monto_descuento = 0.0
            
        elif edad < EDAD_ADULTO:
            precio_base = PRECIO_JOVEN
            tipo_visita = "Joven de (3-17 años)"
            monto_descuento = 0.0
            
        else:  
            precio_base = PRECIO_ADULTOS
         
            print("Tipo de visitante : 1)Estudiante  2)Profesor  3)ninguno")
            opcion = input("Seleccione una opción del 1 al 3 (1 Estudiante, 2 Profesor, 3 Ninguno): ")
            
            if opcion == '1':
                tipo_visita = "Estudiante"
                monto_descuento = precio_base * DESCUENTO_ESTUDIANTE
            elif opcion == '2':
                tipo_visita = "Profesor"
                monto_descuento = precio_base * DESCUENTO_PROFESOR
            else:
                tipo_visita = "Ninguno"
                monto_descuento = precio_base * DESCUENTO_NINGUNO
            
            if edad >= EDAD_MAYOR:
                tipo_visita = tipo_visita + " (Adulto Mayor)"
                descuento_adulto_mayor = precio_base * DESCUENTO_ADULTO_MAYOR
                if descuento_adulto_mayor > monto_descuento:
                    monto_descuento = descuento_adulto_mayor

        monto_paga = precio_base - monto_descuento
        
        total_visitantes_cobrados += 1
        visitante_actual += 1
        total_descuentos_acumulados += monto_descuento
        total_recaudado += monto_paga

        print("Edad:", edad, "|", tipo_visita, "| Base: $", format(precio_base, ".2f"), "| Descuento: $", format(monto_descuento, ".2f"), "| A pagar: $", format(monto_paga, ".2f"), sep="")

    if 'entrada' in locals() and entrada.lower() == 'fin':
        break

    respuesta_final = input("\n¿Quieres repetir el programa? pon 'fin' si quieres cerrar el programa: ")
    if respuesta_final.lower() == 'fin':
        print("Te atendió Diego, vuelve pronto")
        break

print()
print("=== CORTE DE CAJA ===")
print("Visitantes cobrados:", total_visitantes_cobrados)
print("Total en descuentos: $", format(total_descuentos_acumulados, ".2f"), sep="")
print("TOTAL A PAGAR: $", format(total_recaudado, ".2f"), sep="")
```
